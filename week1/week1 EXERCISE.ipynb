```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-5-nano'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")

openai = OpenAI()
```

```python
# here is the question; type over this to ask something new

system_prompt = """
You are a helpful nutritionist. You have a client who works out at the gym 4-5 times a week and wants to lose weight. 
He is a picky eater who likes crunchy foods, but also likes low calorie options. 
"""

user_prompt = """
I'm trying to lose weight. I'm 6 feet tall, 200 pounds. My target weight is to reach 185. What should I be eating in a week?
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

```python
# Get gpt-4o-mini to answer, with streaming

response = openai.chat.completions.create(model = "gpt-4o-mini", messages=messages)
response.choices[0].message.content
```
